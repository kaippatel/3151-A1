    $$p(y_i)={1\over\sqrt{2\pi\sigma^2}}\exp{\left(-{(y_i-(x_i^\top w+b))^2\over 2\sigma^2}\right)}$$
    - Natural log of PDF:
    $$\ln p(y_i)=\ln({1\over\sqrt 2\pi\sigma^2})+\ln\left(\exp{\left( -{(y_i-(x_i^\top w+b))^2\over 2\sigma^2} \right)}\right)\\
                =-{1\over 2}\ln(2\pi\sigma^2)-{(y_i-(x_i^\top w+b))^2\over 2\sigma^2}$$
    - Total log likelihood:
    $$\ln L(w,b,\sigma^2)=\sum_{i=1}^n\left( -{1\over 2}\ln(2\pi\sigma^2)-{(y_i-(x_i^\top w+b))^2\over 2\sigma^2} \right)\\

2. **Explanation for the equivalance to Mean Squared Error (MSE)**:
    
    Maximizing the log-likelihood over $w$ and $b$ for a fixed $\sigma^2$ has different effects on the likelihood formula's terms:
    - The term $-{n\over 2}\ln(2\pi\sigma^2)$ is a constant (with respect to $w$ and $b$) and can be ignored 
    - ${1\over 2\sigma^2}$ is a constant multiplier for the second term (always positive since $\sigma^2>0$)
    - The term $-\sum_{i=1}^n(y_i-(x_i^\top w+b))$ is maximized, which is equivalent to the negative mean squared error.

    Given this is the case, it is clear to see why maximizing the log-likelihood is the same as minimizing the mean squared error, since it maximizes the negative mean squared error. Also, since the constant multiplier ${1\over 2\sigma^2}$ must always be positive, it follows that the second term is always negative.

### 2.2 Implement the core functions (12 marks)
Complete the interfaces below. You may add helpers, but preserve the named functions and return fields.

| Function | Required behaviour |
|:--|:--|

#### `fit_gd` return values

Return a dictionary with `w`, `b`, `loss_history`, `updates`, and `status`.

**Parameters and history.** `updates` is the number of parameter updates actually applied. `loss_history` has length `updates + 1`: the initial objective, followed by the objective after each accepted update. Return the last accepted `w` and `b`, or the initial parameters if no updates were accepted. The last history entry must equal the objective at the returned parameters.

**Status.** Use exactly one of:

- `"completed"`: the full budget of `steps` updates ran. This does **not** imply convergence.
- `"stopped_nonfinite"`: a proposed update produces nonfinite parameters, objective, or gradient.
- `"stopped_bound"`: the proposed parameters, objective, and gradient are finite, but the objective exceeds your documented bound.

**Rejected updates.** Reject an unsafe proposal before applying it. Do not count it in `updates` or append its objective to `loss_history`. Check nonfiniteness before checking the bound. You may use a documented objective bound to stop before overflow.

**Invalid inputs.** Reject invalid inputs or hyperparameters with `ValueError`. The initial parameters, objective, and gradient must be finite; otherwise raise `ValueError` rather than returning a run status. For a valid initial state, `steps=0` returns the initial parameters, a one-entry history, `updates=0`, and `status="completed"`.

Handle constant columns and shape errors sensibly. Do not mutate inputs or silently replace an unstable run by a successful library fit. You need not implement momentum, mini-batches, adaptive optimizers, or early stopping.

There is deliberately no update loop or gradient expression to fill in line by line. Use your derivation and tests.


```python
def fit_standardizer(X: np.ndarray) -> dict:
    """Fit means/scales on X only; population SD; constant-column scale = 1."""
    mean = np.mean(X, axis=0)
    scale = np.std(X, axis=0, ddof=0)
    scale[scale == 0] = 1
    res = {
        'mean': mean,
        'scale': scale
    }
    return res

def apply_standardizer(X: np.ndarray, state: dict) -> np.ndarray:
    """Transform without learning new statistics."""
    return (X - state['mean']) / state['scale']


def loss_and_grad(X: np.ndarray, y: np.ndarray, w: np.ndarray,
                  b: float, lam: float) -> tuple:
    """Return (objective, slope gradient, intercept gradient)."""
    n = X.shape[0]
    y_hat = X @ w + b

    J = (1 / (2 * n)) * np.sum((y_hat - y)**2) + (lam / 2) * np.sum(w**2)
    grad_w = (1 / n) * (y_hat - y) @ X + (lam * w)
    grad_b = np.mean(y_hat - y)

    return J, grad_w, grad_b

def _is_finite_state(
    w: np.ndarray, 
    b: float, 
    J: float, 
    grad_w: np.ndarray, 
    grad_b: float
) -> bool:
    """True if parameters, objective and gradients are all finite, else False."""
    return (
        np.all(np.isfinite(w)) 
        and np.isfinite(b) 
        and np.isfinite(J) 
        and np.all(np.isfinite(grad_w)) 
        and np.isfinite(grad_b)
    )


def fit_gd(X: np.ndarray, y: np.ndarray, lam: float, eta: float,
           steps: int, w0=None, b0: float = 0.0) -> dict:
    """Return w, b, loss_history, updates, status; record initial objective."""
    b = b0
    if w0 is None: 
        w = np.zeros(X.shape[1])
    else: 
        w = np.array(w0)
        
    J, grad_w, grad_b = loss_and_grad(X, y, w, b=b0, lam=lam)

    loss_history = [float(J)]
    updates = 0
    status = 'completed'
    
    BOUND = 1e10
    
    for _ in range(steps):
        new_w = w - eta * grad_w
        new_b = b - eta * grad_b
        new_J, new_grad_w, new_grad_b = loss_and_grad(X, y, new_w, new_b, lam)

        if not _is_finite_state(new_w, new_b, new_J, new_grad_w, new_grad_b):
            status = 'stopped_nonfinite'
            break
        if new_J > BOUND:
            status = 'stopped_bound'
            break

        w, b, J, grad_w, grad_b = new_w, new_b, new_J, new_grad_w, new_grad_b
        updates += 1
        loss_history.append(float(J))

    return {
        'w': w,
        'b': b,
        'loss_history': loss_history,
        'updates': updates,
        'status': status
    }

```

```python
# Fit raw and standardized models with your own GD.
# Build a results table, plot trajectories, and check an independent optimum.
from sklearn.linear_model import Ridge 

def _rmse(y_true, y_pred): 
    """Returns unpenalized RMSE."""
    return float(np.sqrt(np.mean((y_true - y_pred)**2)))

# ------- Fit scalar on training -----------
X_train_fs = train_frame[compact_features].to_numpy(dtype=float)
X_val_fs = valid_frame[compact_features].to_numpy(dtype=float)
y_train = train_frame["energy_kwh"].to_numpy(dtype=float)
y_val = valid_frame["energy_kwh"].to_numpy(dtype=float)

scaler = fit_standardizer(X_train_fs)
representations = {
    "raw": (X_train_fs, X_val_fs),
    "standardized": (
        apply_standardizer(X_train_fs, scaler),
        apply_standardizer(X_val_fs, scaler)
    ),
}

# ------- Config -----------
NUM_STEPS = 250 
LAMBDA = 0.0
b0 = float(y_train.mean())
LEARNING_RATES = {
    "raw": [1e-9, 1e-8, 1e-7, 2e-7, 3e-7], 
    "standardized": [1e-7, 0.01, 0.1, 0.5, 1.0]
}

# ------- Compute curvature (a) per feature -----------
print("------ Curvature (a) and learning rate limits per feature over training rows:")
a_raw = scaler["mean"] ** 2 + scaler["scale"] ** 2 
a_table = pd.DataFrame({
    "feature": compact_features,
    "mean": scaler["mean"],
    "std": scaler["scale"],
    "raw a (curvature)": a_raw,
    "raw eta limit": 2 / a_raw,
    "standardized a (curvature)": 1.0,
    "standardized eta limit": 2.0,
}).sort_values("raw a (curvature)", ascending=True).reset_index(drop=True)
display(a_table)

# ------- Compute independent optimum for each representation -----------
print("------ Independent optimums for Raw and Standardized representations")
J_optimum = {}
for rep, (X_train, X_val) in representations.items(): 
    ref = Ridge(alpha=len(X_train) * LAMBDA).fit(X_train, y_train)
    J_optimum[rep], _, _ = loss_and_grad(X_train, y_train, ref.coef_, ref.intercept_, LAMBDA)
    print(
        f"{rep.capitalize()}:\n"
        f"  - J optimum = {J_optimum[rep]:.2f}\n"
        f"  - RMSE (validation): {_rmse(y_val, X_val @ ref.coef_ + ref.intercept_):.2f}\n"
    )

# ------- Compare learning rates -----------
print(f"------ Gradient descent ({NUM_STEPS} updates) comparison per learning rate:")
records = []
histories = {}
for rep, (X_train, X_val) in representations.items(): 
    for eta in LEARNING_RATES[rep]: 
        fit = fit_gd(X_train, y_train, lam=LAMBDA, eta=eta, steps=NUM_STEPS, b0=b0)
        histories[(rep, eta)] = fit["loss_history"]

        J_final  = fit["loss_history"][-1]
        optimum_gap = (J_final - J_optimum[rep]) / max(1.0, abs(J_optimum[rep]))
        records.append({
            "representation": rep, 
            "eta": eta, 
            "updates": fit["updates"],
            "status": fit["status"],
            "J_final": J_final,
            "optimum_gap": optimum_gap,
            "RMSE_validation": _rmse(y_val, X_val @ fit["w"] + fit["b"]),
        })
display(pd.DataFrame(records))

# ------- Plot overlay loss curves -----------
for rep in representations: 
    J_init = histories[(rep, LEARNING_RATES[rep][0])][0]
    plt.figure()
    for eta in LEARNING_RATES[rep]: 
        plt.plot(histories[(rep, eta)], label=f"eta={eta}")
    plt.axhline(J_optimum[rep], color="black", linestyle="--", label="optimum")
    plt.yscale("log")
    plt.ylim(0.75 * J_optimum[rep], 3 * J_init)
    plt.xlabel("Full-batch update")
    plt.ylabel("Training objective J (log scale)")
    plt.title(f"Gradient descent convergence on {rep} representation (lambda = {LAMBDA})")
    plt.legend()
    plt.show()

# ------- Demonstrate convergence of stable standardized run -----------
ETA_EXTENDED = 0.1
NUM_STEPS_EXTENDED = 2000
X_train_standardized, _ = representations["standardized"]
extended_fit = fit_gd(X_train_standardized, y_train, lam=LAMBDA, eta=ETA_EXTENDED, steps=NUM_STEPS_EXTENDED, b0=b0)
J_final_extended = extended_fit["loss_history"][-1]
_, grad_w, grad_b = loss_and_grad(X_train_standardized, y_train, extended_fit["w"], extended_fit["b"], LAMBDA)
optimum_gap_extended = (J_final_extended - J_optimum['standardized']) / max(1.0, abs(J_optimum['standardized']))

print(f"------- Extended standardized run (eta={ETA_EXTENDED}, {extended_fit['updates']} updates)")
print(f"    - optimum gap = {optimum_gap_extended:.6g}")
print(f"    - gradient norm = {np.linalg.norm(np.append(grad_w, grad_b)):.6g}\n")

# ------- Post-hoc: Determine why standardized eta = 1.0 diverge -----------
print("------ Feature correlations (training rows):")
display(train_frame[compact_features].corr().round(2))
```

```python
# Run the two feature-set sweeps with your GD.
# Store enough information to reconstruct the chosen model without refitting.

# ------- Config -----------
RIDGE_NUM_STEPS = 5000 
RIDGE_LAMBDA_GRID = [0.0, 0.001, 0.01, 0.1, 0.5, 1.0, 5.0, 10.0]
RIDGE_ETA = 0.1 
ACCEPTANCE = 1e-4
FEATURE_SETS = {
    "compact": compact_features, 
    "expanded": expanded_features
}

# ------- Perform regularization sweep on both feature sets-----------
print(f"------- Regularization sweep (eta={RIDGE_ETA}, {RIDGE_NUM_STEPS} updates/run):")
ridge_records = []
ridge_fits = {}
for fs_name, feats in FEATURE_SETS.items(): 
    X_train_fs = train_frame[feats].to_numpy(dtype=float)
    X_val_fs = valid_frame[feats].to_numpy(dtype=float)
    scaler_fs = fit_standardizer(X_train_fs)               
    Z_train = apply_standardizer(X_train_fs, scaler_fs)
    Z_val = apply_standardizer(X_val_fs, scaler_fs)

    for lam in RIDGE_LAMBDA_GRID:
        fit = fit_gd(Z_train, y_train, lam=lam, eta=RIDGE_ETA, steps=RIDGE_NUM_STEPS, b0=b0)

        ref = Ridge(alpha=len(Z_train) * lam).fit(Z_train, y_train)
        J_optimum, _, _ = loss_and_grad(Z_train, y_train, ref.coef_, ref.intercept_, lam)

        J_final = fit["loss_history"][-1]
        optimum_gap = (J_final - J_optimum) / max(1.0, abs(J_optimum))
        ridge_fits[(fs_name, lam)] = {"features": feats, "scaler": scaler_fs, "Z_train": Z_train, "fit": fit}
        ridge_records.append({
            "feature_set": fs_name,
            "lambda": lam,
            "eta": RIDGE_ETA,
            "budget": RIDGE_NUM_STEPS,
            "updates": fit["updates"],
            "status": fit["status"],
            "RMSE_train": _rmse(y_train, Z_train @ fit["w"] + fit["b"]),
            "RMSE_validation": _rmse(y_val, Z_val @ fit["w"] + fit["b"]),
            "w_norm": float(np.linalg.norm(fit["w"])),
            "J_final": J_final,
            "J_optimum": J_optimum,
            "optimum_gap": optimum_gap,
            "did_converge": optimum_gap <= ACCEPTANCE,
        })
ridge_df = pd.DataFrame(ridge_records)
display(ridge_df)

# ------- Plot validation RMSE vs lambda -----------
plt.figure()
for fs_name in FEATURE_SETS:
    rows = ridge_df[ridge_df["feature_set"] == fs_name]
    pos_lambda_rows = rows[rows["lambda"] > 0]     
    line, = plt.semilogx(
        pos_lambda_rows["lambda"], 
        pos_lambda_rows["RMSE_validation"], 
        marker="o", 
        label=fs_name
    )
    plt.axhline(
        rows.loc[rows["lambda"] == 0, "RMSE_validation"].iloc[0],
        color=line.get_color(), 
        linestyle="--", 
        label=f"{fs_name} (lambda = 0)"
    )
plt.xlabel("Ridge lambda (log scale)")
plt.ylabel("Validation RMSE (kWh)")
plt.title("Ridge: validation RMSE vs lambda")
plt.legend()
plt.show()

# ------- Apply runs against selection rule -----------
print("------- Selected run (lowest RMSE among converged runs):")
converged_runs = ridge_df[ridge_df["did_converge"]]
selected_idx = converged_runs["RMSE_validation"].idxmin()
selected_run = ridge_df.loc[selected_idx]
display(ridge_df.loc[[selected_idx]])

# ------- Select model -----------
selected_fit = ridge_fits[(selected_run["feature_set"], selected_run["lambda"])]
_, grad_w, grad_b = loss_and_grad(
    selected_fit["Z_train"], 
    y_train, 
    selected_fit["fit"]["w"], 
    selected_fit["fit"]["b"], 
    selected_run["lambda"]
)

print(f"------- Selected model: {selected_run['feature_set']} features (lambda = {selected_run['lambda']})")
print(f"    - RMSE (validation) = {selected_run['RMSE_validation']:.2f}")
print(f"    - optimum gap = {selected_run['optimum_gap']:.6g}")
print(f"    - gradient norm = {np.linalg.norm(np.append(grad_w, grad_b)):.6g}")

# ------- Store model -----------
selected_model = {
    "features": list(selected_fit["features"]),
    "scaler_mean": selected_fit["scaler"]["mean"].tolist(),
    "scaler_scale": selected_fit["scaler"]["scale"].tolist(),
    "w": selected_fit["fit"]["w"].tolist(),
    "b": float(selected_fit["fit"]["b"]),
    "lambda": float(selected_run["lambda"]),
    "eta": RIDGE_ETA,
    "budget": RIDGE_NUM_STEPS,
    "updates": int(selected_fit["fit"]["updates"]),
    "status": selected_fit["fit"]["status"],
    "optimum_gap": float(selected_run["optimum_gap"]),
    "selection_rule": "lowest validation RMSE among converged runs (optimum gap <= 1e-4)",
}
```
